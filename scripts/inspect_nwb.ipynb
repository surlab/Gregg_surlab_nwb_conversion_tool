# Inspect a converted NWB file

Set `nwb_path` in the next cell to the `.nwb` file you want to inspect (for example after running `run_convert_and_validate_demo.bat`).

Requires `nwbwidgets` in the active conda env (`surlab_nwb_conversion_tool`).

```python
from pathlib import Path

# Edit this path after conversion:
nwb_path = Path("demo_data/SeqBias_AstroBlock_NPX/AstroDREADD25_20240702/AstroDREADD25_20240702.nwb")

if not nwb_path.exists():
    raise FileNotFoundError(f"NWB not found: {nwb_path.resolve()}")

print(nwb_path.resolve())
```

```python
from pynwb import NWBHDF5IO
import nwbwidgets

with NWBHDF5IO(str(nwb_path), mode="r") as io:
    nwbfile = io.read()

nwbwidgets.NWBWidget(nwbfile)
```
